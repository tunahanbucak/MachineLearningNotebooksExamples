```python
with open("myfile.txt",mode="w") as myNewFile:
    myNewFile.write("test4")  # mode w write anlamına gelir fakat burda bi sorun var yazarken dosyadaki diğer şeyleri silerek yazar
```

```python
myNewFile
```

```python
with open("myfile.txt",mode="r") as myFile2:
    myContent = myFile2.read() #mode r read anlamıan gelir
```
