```python
def divideNumber(number):
    return number/2
```

```python
divideNumber(30)
```

```python
myList=[3,5,13,6,21,632,]
```

```python
myResultlist=[]
for num in myList:
    myResultlist.append(divideNumber(num))
print(myResultlist)
```
