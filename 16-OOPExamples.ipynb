```python
class Musician():

    def __init__(self,name):
        self.name = name
        print("musician class")

    def test1(self):
        print("test1")

    def test2(self):
        print("test2")
```

```python
tuna = Musician("tunaaa")
```

```python
tuna.name
```

```python
tuna.test1()
```

```python
tuna.test2()
```

```python
class MusicianPlus(Musician): # burda diğer sınıfın kalıtımını aldık

    def __init__(self,name):
        Musician.__init__(self,name)
        print("musician plus")

    def test3(self):
        print("test3")

    #override üzerine yazma, kalıtım alınan sınıftaki bi methodu değiştirirken yaparız
    def test1(self):
        print("test1test1test1test")
```

```python
atlas = MusicianPlus("atlas")
```

```python
atlas.test1()
```

```python
tuna.test1()
```

```python
atlas.test2()
```

```python
atlas.name = "tunahan buçak"
```

```python
type(atlas)
```

```python
atlas.test3()
```

```python
#POLYMORPHISM
```

```python
class Banana():

    def __init__(self,name):
        self.name = name
    def info(self):
        return f"100 calories {self.name}"

class Apple():

    def __init__(self,name):
        self.name = name

    def info(self):
        return f"150 calories {self.name}"
```

```python
banana = Banana("banana")
```

```python
apple = Apple("apple")
```

```python
banana.info()
```

```python
apple.info()
```

```python
fruitList = [banana,apple]
```

```python
for fruit in fruitList:
    print(fruit.info())
```

```python
#ENCAPSULATION
```

```python
class phone():

    def __init__(self,name,price):
        self.name = name
        self.__price = price # değişkenin dışarıdan müdahale edilememesi için __ kullanarak yazıyor

    def info(self):
        print(f"{self.name} price is: {self.__price}")

    #eğer değiştirmek isityprsak method yapabiliriz
    def changePrice(self,price):
        self.__price = price
```

```python
iphone = phone("iphone 15",600)
```

```python
iphone.info()
```

```python
iphone.__price = 400
```

```python
iphone.info() #değişmedi 
```

```python
iphone.changePrice(300)
```

```python
iphone.info()
```

```python
from abc import ABC, abstractmethod
```

```python
class Car(ABC):

    @abstractmethod
    def maxSpeed(self):
        pass
```

```python
#bu haat verir çünkü Car soyut bi sınıf onu kalıtım için kullanabiliriz  myCar = Car() 
```

```python
class Tesla(Car):

    def maxSpeed(self):
        print("200km")
```

```python
tesla = Tesla()
```

```python
tesla.maxSpeed()
```

```python
class Mercedes(Car):

    def maxSpeed(self):
        print("250km")
```

```python
mercedes = Mercedes()
```

```python
mercedes.maxSpeed()
```

```python
class Fruit():

    def __init__(self,name,calories):
        self.name= name
        self.calories = calories

    def __str__(self):
        return f"{self.name}: {self.calories} calories"

    def __len__(self):
        return self.calories
```

```python
myFruit = Fruit("banana",150)
```

```python
myFruit.calories
```

```python
myFruit.name
```

```python
print(myFruit)
```
